```python
import pandas as pd

CSV_FILE = "incoming_outgoing_Message_Report_2026_08_18_13_05_07.csv"

df = pd.read_csv(CSV_FILE)

df.columns = df.columns.str.strip()

df = df.rename(columns={
    "Visitor Number": "visitor_number",
    "Broadcast Name": "broadcast_name",
    "Attach File": "attach_file",
    "Message": "message",
    "Chat Date Time": "chat_date_time",
    "Customer Read Status": "customer_read_status",
    "Message Charges": "message_charges"
})

df["visitor_number"] = (
    df["visitor_number"]
    .astype(str)
    .str.strip()
)

df["chat_date_time"] = pd.to_datetime(
    df["chat_date_time"],
    errors="coerce"
)

df["chat_date_time"] = df["chat_date_time"].apply(
    lambda x: x.strftime("%Y-%m-%d %H:%M:%S")
    if pd.notna(x)
    else None
)

df["message_charges"] = pd.to_numeric(
    df["message_charges"],
    errors="coerce"
)

df = df.where(pd.notnull(df), None)

records = df.to_dict(orient="records")

existing = (
    supabase
    .table("whatsapp_broadcast_logs")
    .select("id, visitor_number, chat_date_time")
    .execute()
    .data
)

def make_key(visitor, chat_date_time):
    visitor = "" if visitor is None else str(visitor).strip()

    dt = pd.to_datetime(
        chat_date_time,
        errors="coerce"
    )

    if pd.isna(dt):
        dt = ""
    else:
        dt = dt.strftime("%Y-%m-%d %H:%M:%S")

    return visitor, dt

existing_lookup = {}

for row in existing:
    key = make_key(
        row["visitor_number"],
        row["chat_date_time"]
    )

    if key not in existing_lookup:
        existing_lookup[key] = row["id"]

updated = 0
inserted = 0

for record in records:

    key = make_key(
        record["visitor_number"],
        record["chat_date_time"]
    )

    existing_id = existing_lookup.get(key)

    if existing_id is not None:

        supabase \
            .table("whatsapp_broadcast_logs") \
            .update({
                "customer_read_status": record["customer_read_status"]
            }) \
            .eq("id", existing_id) \
            .execute()

        updated += 1

    else:

        supabase \
            .table("whatsapp_broadcast_logs") \
            .insert(record) \
            .execute()

        inserted += 1

print("\nDONE")
print("Updated customer_read_status:", updated)
print("New records inserted:", inserted)
```

```python
for _, row in extra_detail.iterrows():

    din = row.get("DIN")

    if pd.isna(din):
        continue

    din = str(din).strip()

    
    update_data = {
        "reference": "Real Estate Batch 1"
    }

    if pd.notna(row.get("PHONE")):
        update_data["phone"] = str(row['Phone Number']).strip()

    if pd.notna(row.get("EMAIL")):
        update_data["email"] = str(row['Email']).strip()

    
    print(din, update_data.get("phone"), update_data.get("email"))
    if not update_data:
        continue

    (
        supabase
        .table("company_contacts")
        .update(update_data)
        .eq("din", din)
        .execute()
    )
```
